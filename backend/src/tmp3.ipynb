```python
from ag_ui.core import RunAgentInput
from ag_ui_langgraph import LangGraphAgent
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langgraph.checkpoint.memory import MemorySaver
from pydantic_extra_types.uuid_types import uuid7

load_dotenv("../.env")
```

```python
model = "openrouter:liquid/lfm-2.5-2.6b:free"
llm = init_chat_model(model)
agent = create_agent(model, system_prompt="You are helpfully assistance", checkpointer=MemorySaver())
```

```python
agui_agent = LangGraphAgent(name="helpfully agent", graph=agent, emit_raw_events=False)
```

```python
thread_id = str(uuid7())
thread_id
```

```python
from ag_ui.core import UserMessage

input = RunAgentInput(
    thread_id=thread_id,
    run_id=str(uuid7()),
    state=[],
    messages=[
        UserMessage(
            id=str(uuid7()),
            content="What is my name"112
        )],
    tools=[],
    context=[],
    forwardedProps=[]
)
```

```python
seq = [item async for item in agui_agent.run(input)]
```

```python
for mes in seq[-2].messages:
    print(mes)
    print("-"*10)
```

```python
from ag_ui.core import TextMessageContentEvent, EventType
from ag_ui.encoder import EventEncoder

# Create an event
event = TextMessageContentEvent(
    type=EventType.TEXT_MESSAGE_CONTENT,
    message_id="msg_123",
    delta="Hello, world!"
)

# Initialize the encoder
encoder = EventEncoder()

# Encode the event
encoded_event = encoder.encode(event)
print(encoded_event)
# Output: data: {"type":"TEXT_MESSAGE_CONTENT","messageId":"msg_123","delta":"Hello, world!"}\n\n
```
